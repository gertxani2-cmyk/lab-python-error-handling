````python
#2

def initialize_inventory(products):
    inventory = {}
    for product in products:
        valid_quantity = False
        while not valid_quantity:
            try:
                quantity = int(input(f"Enter the quantity of {product} available: "))
                if quantity < 0:
                    raise ValueError("Invalid quantity! Please enter a non-negative value.")
                valid_quantity = True
            except ValueError as error:
                print(f"Error: {error}")
        inventory[product] = quantity
    return inventory

def calculate_total_price(customer_orders):
    total_price = 0
    for product in customer_orders:
        valid_price = False
        while not valid_price:
            try:
                user_input = input(f"Enter price for {product}: ").strip()
                price = float(user_input)
                if price < 0:
                    raise ValueError()
                valid_price = True
            except ValueError:
                print("Oops! Please enter a valid number.")
        total_price += price
    return total_price
            
#3

def get_customer_orders(inventory):
    customer_orders = []
    

    valid_num = False
    while not valid_num:
        try:
            num_orders = int(input("How many products to order? "))
            if num_orders < 0:
                raise ValueError()
            valid_num = True
        except ValueError:
            print("Error: Enter a valid positive number.")
            
    
    for i in range(num_orders):
        valid_product = False
        while not valid_product:
            try:
                product_name = input(f"Product {i+1} name: ").strip()
                
                if product_name not in inventory:
                    raise ValueError("This product does not exist in inventory!")
                if inventory[product_name] == 0:
                    raise ValueError("Sorry, this item is out of stock.")
                    
                customer_orders.append(product_name)
                valid_product = True
            except ValueError as error:
                print(f"Try again: {error}")
                
    return customer_orders

````

````python
products = ["t-shirts", "jeans", "shoes"]
inventory = initialize_inventory(products)
customer_orders = get_customer_orders(inventory)
total_price = calculate_total_price(customer_orders)

print("Total price for the order is:", total_price)
````
